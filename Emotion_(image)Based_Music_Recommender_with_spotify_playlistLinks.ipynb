```python
detector = FER(mtcnn=True)

spotify_playlists = {
    'happy': {
        'title': 'Feel Good Hits',
        'url': 'https://open.spotify.com/playlist/0jrlHA5UmxRxJjoykf7qRY?si=14d265bfcb3f40f5'
    },
    'sad': {
        'title': 'Sad Songs',
        'url': 'https://open.spotify.com/playlist/7gXRobbR7oazZ4PHvWXW33?si=819f16b4dc004c8a'
    },
    'angry': {
        'title': 'Rock Hard',
        'url': 'https://open.spotify.com/playlist/37i9dQZF1DWZryfp6NSvtz?si=eb762dbb31534eb9'
    },
    'surprise': {
        'title': 'Pop Rising',
        'url': 'https://open.spotify.com/playlist/37i9dQZF1DX1QnNyJOBQBv?si=5202a1ad2ca14053'
    },
    'fear': {
        'title': 'Dark & Stormy',
        'url': 'https://open.spotify.com/playlist/0iBjGsWfe7RuZ6yxtJRfsW?si=2f80aea511cd4ca1'
    },
    'disgust': {
        'title': 'Moody Mix',
        'url': 'https://open.spotify.com/playlist/3iPPKuuD1ofFtYbm1TlZnk?si=903a5cfead0540b0'
    },
    'neutral': {
        'title': 'Chill Hits',
        'url': 'https://open.spotify.com/playlist/37i9dQZF1DX4WYpdgoIcn6?si=d14576d6677541e1'
    }
}


```

```python
def detect_emotion_from_image(image_path):
    img = cv2.imread(image_path)
    emotion, score = detector.top_emotion(img)
    return emotion, score

from google.colab import files

uploaded = files.upload()
uploaded_image_path = list(uploaded.keys())[0]

emotion, score = detect_emotion_from_image(uploaded_image_path)
print(f"Detected Emotion: {emotion} (Confidence: {round(score, 2)})")
```

```python
def recommend_playlist(emotion):
    emotion = emotion if emotion in spotify_playlists else 'neutral'
    playlist = spotify_playlists[emotion]

    print(f"\n🎧 Spotify Playlist for Your Mood: {playlist.get('name', 'Unknown')}")
    print(f"▶️ Listen: {playlist.get('url', 'URL Not Found')}")
    return playlist
```

```python
playlist = recommend_playlist(emotion)
```

```python
def show_spotify_playlist(playlist_url):
    if "playlist/" in playlist_url:
        playlist_id = playlist_url.split("playlist/")[1].split("?")[0]
        embed_url = f"https://open.spotify.com/embed/playlist/{playlist_id}"
        display(IFrame(embed_url, width="100%", height="380"))
    else:
        print("Invalid Spotify URL")

show_spotify_playlist(playlist['url'])
```
